```python
import pandas as pd
import numpy as np

train = pd.read_csv("")
sales_test = pd.read_csv('not_encoded_sales_test_.csv')
solution = pd.read_csv('solution.csv')
```

```python
if train.iloc[-1].isna().all():
    train = train.iloc[:-1]

# Creating the 'id' column with the specified format
sales_test['id'] = sales_test.apply(lambda row: f"{row['unique_id']}-{row['year']}_{row['month']:02d}_{row['day']:02d}", axis=1)

# Display the updated DataFrame
print(sales_test)
```

```python
print(len(train) == len(sales_test))  # Returns True if they have the same number of rows
print(f"train rows: {len(train)}, sales_test rows: {len(sales_test)}")  # Print row counts

#add sales_test id to train
train['id'] = sales_test['id']

# Merge and replace sales_hat with predictions
solution = solution.merge(train[['id', 'predictions']], on='id', how='left')
solution['sales_hat'] = solution['predictions']
solution.drop(columns=['predictions'], inplace=True)

# Display the updated solution DataFrame
print(solution)
```
